```python
import os
import pickle

# Ensure this import matches the name of your file containing TRAINING_DATA.
# If you named it dataset.py, change 'sample_data' to 'dataset'
from sample_data import TRAINING_DATA
from nlp_core import preprocess_text, ScratchTfidfVectorizer, ScratchMultinomialNB

def train_and_overwrite_models():
    print("--- STARTING TRAINING PIPELINE ---")
    
    raw_texts = [item[0] for item in TRAINING_DATA]
    labels = [item[1] for item in TRAINING_DATA]
    
    print(f"[+] Loaded {len(raw_texts)} tickets across {len(set(labels))} departments.")

    tokenized_corpus = [preprocess_text(text) for text in raw_texts]
    print("[+] Text preprocessing complete.")

    vectorizer = ScratchTfidfVectorizer()
    X_train = vectorizer.fit_transform(tokenized_corpus)
    print(f"[+] TF-IDF Vectorizer trained. Vocabulary size: {len(vectorizer.vocab)} words.")

    model = ScratchMultinomialNB(alpha=1.0)
    model.fit(X_train, labels)
    print("[+] Naive Bayes model trained with Laplace Smoothing.")

    # Explicitly check for the models directory and overwrite the existing files
    os.makedirs("models", exist_ok=True)
    
    with open('models/custom_tfidf_vectorizer.pkl', 'wb') as f:
        pickle.dump(vectorizer, f)
        
    with open('models/custom_naive_bayes_model.pkl', 'wb') as f:
        pickle.dump(model, f)
        
    print("[+] SUCCESS: Models successfully overwritten in the 'models/' folder!\n")

if __name__ == "__main__":
    train_and_overwrite_models()
```
